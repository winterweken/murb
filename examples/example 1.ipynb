Import modules

```python
from murb_energy_tool import test
from murb_energy_tool import simulation, reporting, utilities, heatbalance
```

**Add windows**

```python
windows_n = simulation.WindowGroup(pct_window_area=0.2, window_azimuth=0, shgc=0.4, shading=0.05)
windows_e = simulation.WindowGroup(pct_window_area=0.2, window_azimuth=90, shgc=0.4, shading=0.05)
windows_s = simulation.WindowGroup(pct_window_area=0.3, window_azimuth=180, shgc=0.4, shading=0.5)
windows_w = simulation.WindowGroup(pct_window_area=0.3, window_azimuth=270, shgc=0.4, shading=0.5)
```

**Specify total enclosure area**

```python
vertical_enclosure = 6650
```

**Run simulation**

```python
sim = simulation.Run(name='Example Building',
                     province='AB',
                     gfa=18279,
                     area_walls=vertical_enclosure*0.66,
                     area_windows=vertical_enclosure*0.33,
                     area_roof=5080,
                     window_groups=[windows_n, windows_e, windows_s, windows_w],
                     setpoint_htg=20,
                     setpoint_clg=24,
                     u_walls=0.210,
                     u_windows=1.9,
                     u_roof=0.138,
                     occupant_density=8,
                     peak_receptacle_load=1,
                     lpd=12.7,
                     fraction_occupied=0.2839,
                     fraction_lighting=0.4586,
                     fraction_receptacle_load=0.4586,
                     service_water_heating_load=30)
```

**Write and print results**

```python
reporting.write_results(sim)
```
